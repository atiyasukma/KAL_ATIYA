# EigenVektor & EigenValue
## Definisi EigenVektor & EigenValue

- Eigenvektor adalah vektor bukan nol yang, ketika diberikan suatu transformasi
linear oleh matriks 𝐴, hasilnya hanya berupa skala (perkalian) dari vektor itu sendiri.Secara matematis, jika 𝑣 adalah eigenvektor dan A, maka:
$$
\begin{array}{cc}
Av=λv
\end{array}
$$


dimana


 $$
\begin{array}{cc}
v ≠ 0
\end{array}
$$

- EigenValue adalah skalar 𝜆 yang merupakan faktor pengali pada eigenvektor 𝑣 saat transformasi linear oleh 𝐴.

Jadi, eigenvalue menunjukkan berapa kali eigenvektor tersebut meregang atau menyusut setelah transformasi.

### Mencari EigenVektor & EigenValue Dengan Polinomial Characteristik urutkan urutan eigen vektor berdasarkan eigen value.

$$det(A−λI)=0$$

$$det(A−λI)=(3−λ)(3−λ)=(3−λ)^2$$



* Langkah 3: Menentukan Eigenvalue



$$(3−λ)^2 =0⟹λ=3$$


Jadi:


$$λ_1 =3,λ_2=3$$
