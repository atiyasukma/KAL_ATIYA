```python
def dot_product(v1, v2):
    return sum(a * b for a, b in zip(v1, v2)) if len(v1) == len(v2) else None

# Contoh penggunaan
vec1 = [2, -3]
vec2 = [4, 1]

hasil = dot_product(vec1, vec2)
print("Dot product:", hasil)
```

```python
from math import sqrt

def panjang_vektor(v):
    return sqrt(sum(x**2 for x in v))

def vektor_satuan(v):
    p = panjang_vektor(v)
    if not p:
        raise ValueError("Vektor nol tidak bisa dikonversi ke vektor satuan.")
    return [x / p for x in v]

# Contoh penggunaan
x = [3, 4]
u = vektor_satuan(x)

print("Vektor satuan:", u)
print("Panjang vektor satuan:", panjang_vektor(u))
```
