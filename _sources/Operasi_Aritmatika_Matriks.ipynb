## DEFINISI MATRIKS
Matriks adalah kumpulan bilangan-bilangan yang disusun secara khusus
dalam bentuk baris dan kolom sehingga membentuk persegi panjang dan bujur
sangkar dimana panjang dan lebarnya ditunjukkan oleh kolom dan baris yang

## OPERASI ARITMATIKA MATRIKS
Merupakan operasi aljabar yang dilakukan pada matriks, seperti penjumlahan, pengurangan, dan perkalian. Operasi ini merupakan bagian dasar dari aljabar linear.
### 1. Penjumlahan Matriks
Dua matriks bisa dijumlahkan jika memiliki ukuran (ordo) yang sama. Elemen-elemen pada posisi yang sama dijumlahkan.

Contoh:
A =
$$
\begin{bmatrix}
& 1 & 2 &  \\
& 3 & 4 & \\
\end{bmatrix}
\quad
$$
B =
$$
\begin{bmatrix}
& 5 & 6 &  \\
& 7 & 8 & \\
\end{bmatrix}
\quad
$$
A - B =
$$
\begin{bmatrix}
& 1+5 & 2+6 \\
& 3+7 & 4+8\\
\end{bmatrix}
\quad
$$
jadi hasilnya
$$
\begin{bmatrix}
& 6 & 8 &  \\
& 10 & 12 & \\
\end{bmatrix}
\quad
$$



```python
# Matriks penjumlahan
def input_matriks():
    return [[int(input(f"Masukkan elemen [{i+1}][{j+1}]: ")) for j in range(2)] for i in range(2)]

def tambah_matriks(A, B):
    return [[A[i][j] + B[i][j] for j in range(2)] for i in range(2)]

print("Masukkan elemen untuk matriks A:")
A = input_matriks()
print("Masukkan elemen untuk matriks B:")
B = input_matriks()

hasil = tambah_matriks(A, B)

print("Hasil penjumlahan matriks:")
for baris in hasil:
    print(baris)
```

### 2. Pengurangan Matriks
Syarat pengurangan matriks sama dengan penjumlahan, yaitu kedua matriks harus memiliki ordo yang sama sehingga bisa dioperasikan.

contoh:

A =
$$
\begin{bmatrix}
& 2 & 4 &  \\
& 3 & 1 & \\
\end{bmatrix}
\quad
$$
B =
$$
\begin{bmatrix}
& 5 & 7 &  \\
& 6 & 2 & \\
\end{bmatrix}
\quad
$$
A - B =
$$
\begin{bmatrix}
& 2-5 & 4-7 \\
& 3-6 & 1-2 \\
\end{bmatrix}
\quad
$$
jadi hasilnya
$$
\begin{bmatrix}
& -3 & -3 &  \\
& -3 & -1 & \\
\end{bmatrix}
\quad
$$

```python
# Pengurangan Matriks
def input_matriks():
    return [[int(input(f"Masukkan elemen [{i+1}][{j+1}]: ")) for j in range(2)] for i in range(2)]

def pengurangan_matriks(A, B):
    return [[A[i][j] - B[i][j] for j in range(2)] for i in range(2)]

print("Masukkan elemen untuk matriks A:")
A = input_matriks()
print("Masukkan elemen untuk matriks B:")
B = input_matriks()

hasil = pengurangan_matriks(A, B)

print("Hasil pengurangan matriks:")
for baris in hasil:
    print(baris)
```

### 3. Perkalian Matriks
Operasi perkalian matriks adalah operasi biner yang menghasilkan matriks baru dari dua matriks yang dikalikan. Perkalian matriks memiliki beberapa jenis, yaitu perkalian skalar dan perkalian dua matriks atau lebih.

contoh:
* Perkalian dua matriks atau lebih

contoh perkalian matriks dibawah menggunakan ordo 2x2 yang memiliki 2 baris dan 2 kolom.

A =
$$
\begin{bmatrix}
& 1 & 5 &  \\
& 6 & 2 & \\
\end{bmatrix}
\quad
$$
B =
$$
\begin{bmatrix}
& 9 & 2 &  \\
& 7 & 8 & \\
\end{bmatrix}
\quad
$$
A x B =
$$
\begin{bmatrix}
& (1×9 + 5×7) & (1×2 + 2×8) \\
& (6×9 + 2×7) & (6×2 + 2×8)  \\
\end{bmatrix}
\quad
$$
jadi hasilnya
$$
\begin{bmatrix}
& 44 & 42 &  \\
& 68 & 28 & \\
\end{bmatrix}
\quad
$$

```python
# Perkalian Matriks (perkalian dua matriks atau lebih)
def input_matriks():
    return [[int(input(f"Masukkan elemen baris {i+1}, kolom {j+1}: ")) for j in range(2)] for i in range(2)]

def kali_matriks(A, B):
    return [[sum(A[i][k] * B[k][j] for k in range(2)) for j in range(2)] for i in range(2)]

print("Masukkan elemen untuk Matriks A:")
A = input_matriks()
print("Masukkan elemen untuk Matriks B:")
B = input_matriks()

hasil = kali_matriks(A, B)

print("Hasil perkalian matriks:")
for baris in hasil:
    print(baris)
```

*   Perkalian Skalar Matriks

Perkalian ini dilakukan dengan mengalikan setiap elemen dalam matriks dengan sebuah bilangan skalar.

contoh:
jika A =
$$
\begin{bmatrix}
& 4 & 7 &  \\
& 5 & 8 & \\
\end{bmatrix}
\quad
$$
dan skalar  k = 3, maka:
              
kA = 3 x $$
\begin{bmatrix}
& 4 & 7 &  \\
& 5 & 8 & \\
\end{bmatrix}
\quad
$$
jadi hasilnya
$$
\begin{bmatrix}
& 12 & 21 &  \\
& 15 & 24 & \\
\end{bmatrix}
\quad
$$

```python
# perkalian skalar matriks
def kali_skalar_matriks(k, A):
    return [[k * A[i][j] for j in range(len(A[0]))] for i in range(len(A))]

def input_matriks():
    return [[int(input(f"Masukkan elemen baris {i+1}, kolom {j+1}: ")) for j in range(2)] for i in range(2)]

k = int(input("Masukkan skalar: "))
print("Masukkan elemen untuk Matriks A:")
A = input_matriks()

hasil = kali_skalar_matriks(k, A)

print("Hasil perkalian skalar dengan matriks:")
for baris in hasil:
    print(baris)
```

## Determinan matriks
Merupakan suatu nilai skalar yang dapat dihitung dari elemen-elemen suatu matriks persegi yang biasanya matriks dengan jumlah baris dan kolom yang sama.
digunakan untuk mementukan apakah suatu matriks memiliki invers (det (A)=/ 0).
 * sifat sifat penting determinan
1. Determinan Matriks identitas:
det(I)=1, dimana I adalah matriks identitas
2. Baris atau kolom nol:
Jika ada baris atau kolom yang semuanya nol, maka det(A)=0
3. pertukaran baris/kolom;
Jika dua baris atau kolom dipertukarkan, determinan berubah tanda
4. Kelipatan baris/kolom:
Jika satu baris atau kolom dikalikan dengan skalar k, maka determinan menjadi k det (A)
5. matriks singular:
Jika det(A)=0, maka matriks A disebut singular(Tidak memiliki invers) jika non singular maka matriks nya buka 0, tapi -1 memiliki invers (sebalik nya dari matriks singular)
* contoh matriks 2x2:
$$
\begin{bmatrix}
& a & b &  \\
& c & d & \\
\end{bmatrix}
\quad
$$

maka determinannya adalah:
det(A) = ad−bc
* contoh matriks 3x3:
$$
\begin{bmatrix}
& a & b & c & \\
& d & e & f & \\
& g & h & i & \\
\end{bmatrix}
\quad
$$

maka determinannya adalah:
det(A) = a(ei−fh)−b(di−fg)+c(dh−eg)




### Minor Matriks
Merupakan bagian matriks yang tersisa setelah menghilangkan baris dan kolom yang berisi elemen tertentu.
* contoh matriks 3x3:

A =
$$
\begin{bmatrix}
& 1 & 2 & 3 &\\
& 4 & 5 & 6 &\\
& 7 & 8 & 9 &\\
\end{bmatrix}
\quad
$$
Minor untuk elemen A $M_{23}$ = 6

$M_{23}$:
$$
\begin{bmatrix}
& 1 & 2 &  \\
& 7 & 8 & \\
\end{bmatrix}
\quad
$$
$M_{23}$ ​= (1×8)−(2×7) = 8−14 = −6



### Kofaktor Matriks
Merupakan kofaktor dari suatu elemen dalam matriks adalah minor dari elemen tersebut yang dikalikan dengan faktor tanda berdasarkan posisinya dalam matriks.
* Rumus Kofaktor
$
C_{ij} = (-1)^{i+j} M_{ij}
$
* contoh matriks 3x3:
* menentukan minor

A =
$
\begin{bmatrix}

& 4 & 5 & 6 &\\

\quad
$

$
M_{23}:
$
$
\begin{bmatrix}
& 1 & 2 &  \\
& 7 & 8 & \\
\end{bmatrix}
\quad
$
$M_{23}$ ​= (1×8)−(2×7) = 8−14 = −6
* lalu masukkan rumus kofaktor
$$
C_{ij} = (-1)^{i+j} M_{ij}
$$
* masukkan angka

$$

$$

#### Hubungan Matriks Minor & Kofaktor
* Minor $(M_{ij})$ adalah determinan dari submatriks yang diperoleh dengan menghapus baris ke-i dan kolom ke-j.
* Kofaktor $(C_{ij})$ adalah minor yang dikalikan dengan faktor tanda $(−1)^{i+j}$ dan digunakan dalam perhitungan determinan, invers matriks, serta ekspansi kofaktor.
